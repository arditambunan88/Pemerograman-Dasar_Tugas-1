```python
# Data: energi terkirim harian unit PLTU 330 MW, 1–30 November 2026 (MWh)
energi = [
    5576, 6098, 6078, 6095, 6074, 6077, 5564, 5552, 6100, 6083,   # hari 1–10
    6097, 6089, 6066, 5533, 5566, 6097, 3592, 6059, 6100, 6098,   # hari 11–20
    5565, 5571, 6103, 6135, 6065, 6071, 6086, 5577, 5575, 6062,   # hari 21–30
]
hari = list(range(1, 31))          # array paralel: hari[i] = nomor hari untuk energi[i]
JUMLAH_PERINGKAT = 3               # tiga hari tertinggi / terendah
JENDELA = 7                        # rata-rata bergerak 7 hari
BATAS_PERSEN = 97                  # batas rata-rata bergerak: 97% rata-rata bulanan
HARI_PER_MINGGU = 7
GARIS = "=" * 52

# Input target harian (dengan try/except dari Minggu 5)
target = None
while target is None:
    try:
        target = float(input("Masukkan target energi harian (MWh): "))
        if target <= 0:
            print("Target harus lebih dari 0.")
            target = None
    except ValueError:
        print("Masukan tidak valid, ketik angka (contoh: 5800).")
```

```python
# 1) Total dan rata-rata (akumulator)

total = 0
for nilai in energi:
    total = total + nilai
rata_bulan = total / len(energi)

print("\n[1] RINGKASAN")
print(f"    {'Total energi':<22}: {total:>10,} MWh")
print(f"    {'Rata-rata harian':<22}: {rata_bulan:>10,.1f} MWh/hari")
```

```python
# 2) Tiga hari tertinggi dan tiga terendah
#    Cari posisi ekstrem berulang kali, lewati indeks yang sudah terpilih.
#    Dengan '>' / '<', jika nilainya sama, hari yang lebih awal dipilih dulu.

tertinggi = []                     # berisi INDEKS, bukan nilai
for _ in range(JUMLAH_PERINGKAT):
    i_maks = None
    for i in range(len(energi)):
        if i in tertinggi:
            continue
        if i_maks is None or energi[i] > energi[i_maks]:
            i_maks = i
    tertinggi.append(i_maks)

terendah = []
for _ in range(JUMLAH_PERINGKAT):
    i_min = None
    for i in range(len(energi)):
        if i in terendah:
            continue
        if i_min is None or energi[i] < energi[i_min]:
            i_min = i
    terendah.append(i_min)

print(f"\n[2] {JUMLAH_PERINGKAT} HARI TERTINGGI")
for p in range(len(tertinggi)):
    i = tertinggi[p]
    print(f"    {p + 1}. Hari ke-{hari[i]:<4} {energi[i]:>8,} MWh")

print(f"\n    {JUMLAH_PERINGKAT} HARI TERENDAH")
for p in range(len(terendah)):
    i = terendah[p]
    print(f"    {p + 1}. Hari ke-{hari[i]:<4} {energi[i]:>8,} MWh")
```

```python
# 3) Hari di bawah target (penyaringan)

hari_kurang = []
for i in range(len(energi)):
    if energi[i] < target:
        hari_kurang.append(hari[i])

print("\n[3] PERBANDINGAN DENGAN TARGET")
print(f"    {'Target harian':<22}: {target:>10,.0f} MWh")
print(f"    {'Hari di bawah target':<22}: {len(hari_kurang):>10} dari {len(energi)} hari")
if len(hari_kurang) > 0:
    print(f"    {'Yaitu hari ke-':<22}: {hari_kurang}")
```

```python
# 4) Rata-rata bergerak 7 hari di bawah 97% rata-rata bulanan
#    Jendela penuh pertama berakhir di indeks 6 (hari ke-7).

batas = BATAS_PERSEN / 100 * rata_bulan
hari_waspada = []                  # pasangan [nomor hari, rata-rata bergerak]
for i in range(JENDELA - 1, len(energi)):
    jendela = energi[i - JENDELA + 1 : i + 1]
    rata_gerak = sum(jendela) / JENDELA
    if rata_gerak < batas:
        hari_waspada.append([hari[i], rata_gerak])

print(f"\n[4] RATA-RATA BERGERAK {JENDELA} HARI < {BATAS_PERSEN}% RATA-RATA BULANAN")
print(f"    {'Batas':<22}: {batas:>10,.1f} MWh/hari")
if len(hari_waspada) == 0:
    print("    Tidak ada hari yang ditandai.")
else:
    for data in hari_waspada:
        print(f"    Hari ke-{data[0]:<4} rata-rata {JENDELA} hari: {data[1]:>8,.1f} MWh/hari")
```

```python
# 5) Total energi per minggu (minggu 1–4) + sisa hari

total_mingguan = []
jumlah_minggu_penuh = len(energi) // HARI_PER_MINGGU      # 30 // 7 = 4
for m in range(jumlah_minggu_penuh):
    awal = m * HARI_PER_MINGGU
    akhir = awal + HARI_PER_MINGGU
    total_mingguan.append(sum(energi[awal:akhir]))
awal_sisa = jumlah_minggu_penuh * HARI_PER_MINGGU
total_sisa = sum(energi[awal_sisa:])

print("\n[5] TOTAL ENERGI PER MINGGU")
for m in range(len(total_mingguan)):
    awal = m * HARI_PER_MINGGU + 1
    akhir = awal + HARI_PER_MINGGU - 1
    label = f"Minggu {m + 1} (hari {awal}–{akhir})"
    print(f"    {label:<26}: {total_mingguan[m]:>8,} MWh")
if awal_sisa < len(energi):
    label = f"Sisa (hari {awal_sisa + 1}–{len(energi)})"
    print(f"    {label:<26}: {total_sisa:>8,} MWh")
print(f"    {'Total (cek)':<26}: {sum(total_mingguan) + total_sisa:>8,} MWh")
```
