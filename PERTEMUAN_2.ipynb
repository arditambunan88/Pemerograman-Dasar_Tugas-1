```python
# Jawaban Latihan 1 — Membuat variabel
tegangan_surya = 18
arus_surya = 6
jenis_sel = "Monocrystalline"

print("Tegangan:", tegangan_surya, "volt")
print("Arus:", arus_surya, "ampere")
print("Jenis Sel:", jenis_sel)
```

```python
# Jawaban Latihan 2 — Tipe primitif & type()
jumlah_baterai = 4              # int
suhu_panel = 36.5               # float
kipas_aktif = True              # bool
merek_inverter = "Schneider"    # str

print("Nilai:", jumlah_baterai, "| Tipe:", type(jumlah_baterai))
print("Nilai:", suhu_panel, "| Tipe:", type(suhu_panel))
print("Nilai:", kipas_aktif, "| Tipe:", type(kipas_aktif))
print("Nilai:", merek_inverter, "| Tipe:", type(merek_inverter))
```

```python
# Jawaban Latihan 3 — Memilih tipe yang tepat
nama_unit = "PLTS Nunukan"
jumlah_panel = 240
tegangan_keluaran = 48.5
sedang_beroperasi = True

print("Nama unit:", nama_unit, "| Tipe:", type(nama_unit))
print("Jumlah panel:", jumlah_panel, "| Tipe:", type(jumlah_panel))
print("Tegangan keluaran:", tegangan_keluaran, "| Tipe:", type(tegangan_keluaran))
print("Sedang beroperasi:", sedang_beroperasi, "| Tipe:", type(sedang_beroperasi))
```

```python
# Jawaban Latihan 4 — List
tegangan_harian = [217.5, 219.0, 220.3, 218.7, 221.1, 216.9]

# (a) Cetak pembacaan pertama dan terakhir
print("Pembacaan pertama:", tegangan_harian[0])
print("Pembacaan terakhir:", tegangan_harian[-1])

# (b) Tambahkan satu pembacaan baru 222.0 dengan append()
tegangan_harian.append(222.0)

# (c) Cetak jumlah total data dengan len()
print("Jumlah total data:", len(tegangan_harian))
print("List terbaru:", tegangan_harian)
```

```python
# Jawaban Latihan 5 — Dictionary
motor_listrik = {
    "nama": "Motor Induksi A",
    "daya_kw": 15.5,
    "tegangan_v": 380,
    "rpm": 1450
}

# Cetak salah satu nilai lewat key-nya
print("RPM awal:", motor_listrik["rpm"])

# Ubah nilai rpm
motor_listrik["rpm"] = 1500

# Cetak ulang seluruh dictionary
print("Dictionary motor listrik setelah diubah:", motor_listrik)
```

```python
# Jawaban Latihan 6 — Gabungan
pembangkit = [
    {"nama": "Unit-1", "kapasitas_MW": 100, "status": "operasi"},
    {"nama": "Unit-2", "kapasitas_MW": 150, "status": "operasi"},
    {"nama": "Unit-3", "kapasitas_MW": 80, "status": "pemeliharaan"},
]

# Cetak nama unit kedua
print("Nama unit kedua:", pembangkit[1]["nama"])

# Ubah status salah satu unit
pembangkit[0]["status"] = "standby"

# Tambahkan satu unit baru ke list dengan append()
pembangkit.append({"nama": "Unit-4", "kapasitas_MW": 120, "status": "operasi"})

# Cetak seluruh data
print("Seluruh data pembangkit:")
for unit in pembangkit:
    print(unit)
```
